Use somente apelidos. Os prêmios são fictícios. A credencial privada fica nos Secrets do Colab, fora do código e das saídas. Não compartilhe a chave nem a coloque no site.

1. No Supabase, abra Settings → API Keys e copie a Secret key (ou service_role legada).
2. No Colab, abra o ícone 🔑 → adicionar secret chamado SUPABASE_KEY e ative o acesso deste notebook.
3. Execute as células abaixo. A segunda abre o menu do jogo.
4. Ao terminar a partida, atualize o site e compare apelido, prêmio e duração.

O script SQL de permissões precisa estar aplicado ao Supabase. A chave pública do site não é a chave de execução do jogo.

```python
import os
from getpass import getpass
from google.colab import userdata

os.environ["SUPABASE_URL"] = "https://mkgbeemjhvpvjmqbwtfl.supabase.co"
try:
    chave = userdata.get("SUPABASE_KEY")
except Exception:
    chave = getpass("Chave privada do Supabase (não aparece na tela): ")
if not chave or chave.startswith("sb_publishable_"):
    raise ValueError("Configure SUPABASE_KEY com a chave privada de execução, não a publishable do site.")
os.environ["SUPABASE_KEY"] = chave.strip()
del chave
print("Configuração carregada; chave não exibida.")
```
